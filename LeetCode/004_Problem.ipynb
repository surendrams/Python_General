# LeetCode #4: Median of Two Sorted Arrays

We need an algorithm with O(log(min(m, n))) complexity.
The best approach is to use binary search on the smaller array.

---

### Example 1:
Input: nums1 = [1,3], nums2 = [2]
Output: 2.00000
Explanation: merged array = [1,2,3] and median is 2.

### Example 2:
Input: nums1 = [1,2], nums2 = [3,4]
Output: 2.50000
Explanation: merged array = [1,2,3,4] and median is (2 + 3) / 2 = 2.5.


### Key Idea:
We want to partition both arrays such that:
Left partition contains half of the total elements (or half + 1 if odd).
All elements in left partition ≤ all elements in right partition.
Median is then computed based on max of left partition and min of right partition

### Algorithm (Binary Search)
- Ensure nums1 is the smaller array (m <= n).
- Use binary search on nums1:
    - i is partition index in nums1
    - j = (m + n + 1) // 2 - i is partition index in nums2
- Check conditions:
    - If nums1[i-1] <= nums2[j] and nums2[j-1] <= nums1[i], correct partition found.
    - Else adjust binary search range.
- Median:
    - If (m + n) is odd → max(left partition)
    - Else → average of max(left partition) and min(right partition)

```python
def findMedianSortedArrays(nums1, nums2):
    m, n = len(nums1), len(nums2)
    
    # Ensure nums1 is smaller
    if m > n:
        return findMedianSortedArrays(nums2, nums1)
    
    left, right = 0, m
    half_len = (m + n + 1) // 2
    
    while left <= right:
        i = (left + right) // 2
        j = half_len - i
        
        max_left1 = float('-inf') if i == 0 else nums1[i - 1]
        min_right1 = float('inf') if i == m else nums1[i]
        
        max_left2 = float('-inf') if j == 0 else nums2[j - 1]
        min_right2 = float('inf') if j == n else nums2[j]
        
        if max_left1 <= min_right2 and max_left2 <= min_right1:
            if (m + n) % 2 == 1:
                return max(max_left1, max_left2)
            else:
                return (max(max_left1, max_left2) + min(min_right1, min_right2)) / 2
        elif max_left1 > min_right2:
            right = i - 1
        else:
            left = i + 1
```
