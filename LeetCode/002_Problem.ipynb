# LeetCode #2: Add Two Numbers problem

You are given two non-empty linked lists representing two non-negative integers. The digits are stored in reverse order, and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list. 

You may assume the two numbers do not contain any leading zero, except the number 0 itself.

---

### 1. Input Format
The problem statement says: The digits are stored in reverse order, and each node contains a single digit.
That means:
- Number 342 is stored as [2 → 4 → 3].
- Number 465 is stored as [5 → 6 → 4].
    So the least significant digit (ones place) is already at the head of the list.
### 2. How the Addition Works
When we add numbers by hand, we also start from the ones place:
    342
   +465

    807

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def addTwoNumbersForward(l1: ListNode, l2: ListNode) -> ListNode:
    dummy = ListNode()
    curr = dummy
    carry = 0

    while l1 or l2 or carry:
        val1 = l1.val if l1 else 0
        val2 = l2.val if l2 else 0
        total = val1 + val2 + carry
        
        carry = total // 10
        curr.next = ListNode(total % 10)
        curr = curr.next

        if l1: l1 = l1.next
        if l2: l2 = l2.next

    return dummy.next
    
```

```python
l1 = ListNode(3)
l1.next = ListNode(4)
l1.next.next = ListNode(2)

l2 = ListNode(4)
l2.next = ListNode(6)
l2.next.next = ListNode(5)

# Call reverse-order add function
result_node = addTwoNumbersForward(l1, l2)

# Print result
while result_node:
    print(result_node.val)
    result_node = result_node.next
```
