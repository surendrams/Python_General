```python
# Single Linked List

class SinglyNode:

    def __init__(self, val, next=None) -> None:
        self.val = val
        self.next = next

    def __str__(self):
        return self.val
```

```python
# Single Lined List Output
Head = SinglyNode(0)
vara = SinglyNode(11)
varb = SinglyNode(23)
varc = SinglyNode(32)
Head.next = vara
vara.next = varb
varb.next = varc
curr = Head
while curr:
    print(f"{curr.val}")
    curr = curr.next
```

```python
# Double Linked List

class DoublyNode:

    def __init__(self, val):
        self.val = val
        self.next = None
        self.prev = None

    def __str__(self) -> str:
        return str(self.val)
```

```python
def display(head):
    curr = head
    while curr:
        print(curr.val, end=" <-> ")
        curr = curr.next
```

```python
def insert_at_beginning(head, val):
    new_node = DoublyNode(val)
    new_node.next = head
    if head:
        head.prev = new_node

    return new_node
```

```python
def insert_at_end(head, val):
    new_node = DoublyNode(val)
    if head is None:
        return new_node

    curr = head
    while curr.next:
        curr = curr.next

    curr.next = new_node
    new_node.prev = curr

    return head
```

```python
head = DoublyNode(100)
head = insert_at_beginning(head, 90)
head = insert_at_beginning(head, 80)
head = insert_at_end(head, 1000)
display(head)

```
