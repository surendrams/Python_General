```python
import json
from pathlib import Path

from pymongo import MongoClient

mongo_uri = "mongodb://localhost:27017/"
db_name = "lumi_db"
collection_name = "certification_questions"
json_path = Path.cwd() / "questions_ALL.json"

with json_path.open("r", encoding="utf-8") as f:
    questions = json.load(f)

if not isinstance(questions, list):
    raise ValueError("questions_ALL.json must contain a JSON array of question documents")

client = MongoClient(mongo_uri)
collection = client[db_name][collection_name]

# Optional: avoid duplicate inserts if docs already have a stable id field.
collection.delete_many({})

if questions:
    result = collection.insert_many(questions)
    print(f"Inserted {len(result.inserted_ids)} documents into {db_name}.{collection_name}")
else:
    print("questions_ALL.json is empty; nothing was inserted")

print(f"Source file: {json_path}")
print(f"Document count: {collection.count_documents({})}")
```
